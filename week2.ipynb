## Property Price Prediction using Linear Regression

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

### 1. Load the property dataset

```python
df = pd.read_csv('property.csv')
```

### 2. Define target and full feature matrix

```python
X_full = df[
    [
        'Square_Footage',
        'Num_Bedrooms',
        'Num_Bathrooms',
        'Year_Built',
        'Lot_Size',
    ]
]
y = df['Price']
```

### 3. Train-Test Split (80% Train, 20% Test)

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_full, y, test_size=0.2, random_state=42
)
```

### 4. Define Feature Combinations for Multiple Linear Regression Models

```python
feature_combinations = {
    'Model 1 (Baseline)': ['Square_Footage'],
    'Model 2 (3 Features)': [
        'Square_Footage',
        'Num_Bedrooms',
        'Num_Bathrooms',
    ],
    'Model 3 (4 Features)': [
        'Square_Footage',
        'Num_Bedrooms',
        'Num_Bathrooms',
        'Year_Built',
    ],
    'Model 4 (All 5 Features)': [
        'Square_Footage',
        'Num_Bedrooms',
        'Num_Bathrooms',
        'Year_Built',
        'Lot_Size',
    ],
}
```

### 5. Model Training & Performance Evaluation

```python
model_performance = []

for model_name, features in feature_combinations.items():
  # Fit Linear Regression Model
  model = LinearRegression()
  model.fit(X_train[features], y_train)

  # Generate Predictions
  y_pred = model.predict(X_test[features])

  # Compute Metrics
  r2 = r2_score(y_test, y_pred)
  mae = mean_absolute_error(y_test, y_pred)
  rmse = np.sqrt(mean_squared_error(y_test, y_pred))

  model_performance.append({
      'Model': model_name,
      'Features Used': ', '.join(features),
      'R-Squared (R²)': round(r2, 6),
      'MAE ($)': round(mae, 2),
      'RMSE ($)': round(rmse, 2),
  })
```

### 6. Display Model Performance

```python
# Convert evaluation results to DataFrame
results_df = pd.DataFrame(model_performance)
print(results_df.to_string(index=False))
```
